```python

class Config:
    """
    Central configuration for the Fuzzy Logic CNN Enhancement System.
    Import this class in every other notebook to keep settings consistent.
    """

    # --- Image preprocessing ---
    IMAGE_SIZE       = (128, 128)   # Resize target (H, W)
    NORMALIZE_MEAN   = [0.5]        # For grayscale; use [0.5, 0.5, 0.5] for RGB
    NORMALIZE_STD    = [0.5]

    # --- CNN training ---
    BATCH_SIZE       = 32
    LEARNING_RATE    = 1e-3
    NUM_EPOCHS       = 30
    WEIGHT_DECAY     = 1e-4
    DROPOUT_RATE     = 0.3

    # --- CNN output range (physical units) ---
    BRIGHTNESS_RANGE = (0.5, 2.0)   # alpha multiplier for brightness
    CONTRAST_RANGE   = (0.5, 2.0)   # alpha multiplier for contrast
    GAMMA_RANGE      = (0.4, 2.5)   # gamma correction exponent

    # --- Fuzzy universe of discourse ---
    BRIGHTNESS_UNIVERSE = (0.0, 1.0)   # normalised mean intensity
    CONTRAST_UNIVERSE   = (0.0, 1.0)   # normalised std deviation
    ADJUSTMENT_UNIVERSE = (-1.0, 1.0)  # output: negative=decrease, positive=increase

    # --- File paths ---
    MODEL_SAVE_PATH  = "cnn_enhancer.pth"
    SAMPLE_IMAGE     = "sample.jpg"     # default test image

print("Config loaded successfully.")
print(f"   Image size      : {Config.IMAGE_SIZE}")
print(f"   Batch size      : {Config.BATCH_SIZE}")
print(f"   Learning rate   : {Config.LEARNING_RATE}")
print(f"   Num epochs      : {Config.NUM_EPOCHS}")
print(f"   Model save path : {Config.MODEL_SAVE_PATH}")
```

```python
# ── Save Config to disk so other notebooks can import it ─────────────────────
config_source = '''
class Config:
    IMAGE_SIZE           = (128, 128)
    NORMALIZE_MEAN       = [0.5]
    NORMALIZE_STD        = [0.5]
    BATCH_SIZE           = 32
    LEARNING_RATE        = 1e-3
    NUM_EPOCHS           = 30
    WEIGHT_DECAY         = 1e-4
    DROPOUT_RATE         = 0.3
    BRIGHTNESS_RANGE     = (0.5, 2.0)
    CONTRAST_RANGE       = (0.5, 2.0)
    GAMMA_RANGE          = (0.4, 2.5)
    BRIGHTNESS_UNIVERSE  = (0.0, 1.0)
    CONTRAST_UNIVERSE    = (0.0, 1.0)
    ADJUSTMENT_UNIVERSE  = (-1.0, 1.0)
    MODEL_SAVE_PATH      = "cnn_enhancer.pth"
    SAMPLE_IMAGE         = "sample.jpg"
'''

with open("config.py", "w") as f:
    f.write(config_source)

print("config.py written to disk — other notebooks will import from here.")
```
