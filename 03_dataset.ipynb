# Module 3 — Synthetic Dataset Generator

## Why synthetic data?

We don't always have paired **(degraded → reference)** image datasets.  
So we **create** them programmatically:

```
Reference image  ──► apply random degradation ──► Degraded image
                                                        │
                  compute ideal enhancement params ◄────┘
                                 │
                    CNN learns to predict these params
```

**Degradation types:**
- `dark`         — multiply pixels by 0.3–0.7
- `bright`       — multiply pixels by 1.4–2.5
- `low_contrast` — compress dynamic range
- `noise`        — add Gaussian noise


````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch.utils.data import Dataset

from config            import Config
from feature_extractor import extract_features, features_to_vector

print("Imports successful.")
````

````python
#Ideal parameter computation

def compute_ideal_params(degraded: np.ndarray, reference: np.ndarray) -> np.ndarray:
    """
    Heuristically compute the brightness / contrast / gamma factors
    that would bring `degraded` closer to `reference`.

    Returns
    -------
    np.ndarray of shape (3,):
        [brightness_factor_norm, contrast_factor_norm, gamma_norm]
    All values normalised to [0, 1] for the CNN sigmoid output layer.
    """
    ref_mean = np.mean(reference) / 255.0 + 1e-8
    deg_mean = np.mean(degraded)  / 255.0 + 1e-8
    ref_std  = np.std(reference)  / 255.0 + 1e-8
    deg_std  = np.std(degraded)   / 255.0 + 1e-8

    brightness_factor = np.clip(ref_mean / deg_mean,
                                *Config.BRIGHTNESS_RANGE)
    contrast_factor   = np.clip(ref_std  / deg_std,
                                *Config.CONTRAST_RANGE)
    gamma = np.clip(np.log(0.5 + 1e-8) / np.log(deg_mean + 1e-8),
                    *Config.GAMMA_RANGE)

    def norm(v, lo, hi):
        return (v - lo) / (hi - lo + 1e-8)

    return np.array([
        norm(brightness_factor, *Config.BRIGHTNESS_RANGE),
        norm(contrast_factor,   *Config.CONTRAST_RANGE),
        norm(gamma,             *Config.GAMMA_RANGE),
    ], dtype=np.float32)


print("compute_ideal_params() defined.")
````

````python
#PyTorch Dataset class

class SyntheticEnhancementDataset(Dataset):

    def __init__(self, num_samples: int = 5000,
                 img_size: tuple = Config.IMAGE_SIZE):
        self.num_samples = num_samples
        self.img_size    = img_size

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        reference         = self._make_reference()
        degraded, mode    = self._apply_degradation(reference)
        feats             = extract_features(degraded)
        feat_vec          = features_to_vector(feats)          # (260,)
        targets           = compute_ideal_params(degraded, reference)  # (3,)
        return (torch.tensor(feat_vec, dtype=torch.float32),
                torch.tensor(targets,  dtype=torch.float32))

    def _make_reference(self) -> np.ndarray:
        """Generate a naturalistic grayscale image."""
        H, W  = self.img_size
        base  = np.random.randint(60, 200, (H, W), dtype=np.uint8)
        noise = np.random.normal(0, 20, (H, W)).astype(np.float32)
        img   = np.clip(base.astype(np.float32) + noise, 0, 255).astype(np.uint8)
        if np.random.rand() < 0.5:
            k = np.random.choice([3, 5])
            img = cv2.GaussianBlur(img, (k, k), 0)
        return img

    def _apply_degradation(self, img: np.ndarray):
        """Apply one of four degradation types randomly."""
        mode = np.random.choice(["dark", "bright", "low_contrast", "noise"])

        if mode == "dark":
            f        = np.random.uniform(0.3, 0.7)
            degraded = np.clip(img.astype(np.float32) * f, 0, 255).astype(np.uint8)
        elif mode == "bright":
            f        = np.random.uniform(1.4, 2.5)
            degraded = np.clip(img.astype(np.float32) * f, 0, 255).astype(np.uint8)
        elif mode == "low_contrast":
            degraded = cv2.convertScaleAbs(img, alpha=0.4, beta=80)
        else:  # noise
            noise    = np.random.normal(0, 30, img.shape).astype(np.float32)
            degraded = np.clip(img.astype(np.float32) + noise, 0, 255).astype(np.uint8)

        return degraded, mode


print("SyntheticEnhancementDataset class defined.")
````

````python
#visualise sample degraded/reference pairs

ds   = SyntheticEnhancementDataset(num_samples=100)
ref  = ds._make_reference()

fig, axes = plt.subplots(2, 4, figsize=(14, 5))
modes = ["dark", "bright", "low_contrast", "noise"]

for col, mode_name in enumerate(modes):
    # Override degradation mode for demo
    ds_tmp = SyntheticEnhancementDataset(num_samples=1)
    # Manually apply each degradation type
    if mode_name == "dark":
        deg = np.clip(ref.astype(np.float32) * 0.4, 0, 255).astype(np.uint8)
    elif mode_name == "bright":
        deg = np.clip(ref.astype(np.float32) * 2.0, 0, 255).astype(np.uint8)
    elif mode_name == "low_contrast":
        deg = cv2.convertScaleAbs(ref, alpha=0.4, beta=80)
    else:
        noise = np.random.normal(0, 40, ref.shape).astype(np.float32)
        deg   = np.clip(ref.astype(np.float32) + noise, 0, 255).astype(np.uint8)

    params = compute_ideal_params(deg, ref)

    # Reference (top row)
    axes[0, col].imshow(ref, cmap="gray", vmin=0, vmax=255)
    axes[0, col].set_title(f"Reference\nmean={ref.mean()/255:.2f}", fontsize=9)
    axes[0, col].axis("off")

    # Degraded (bottom row)
    axes[1, col].imshow(deg, cmap="gray", vmin=0, vmax=255)
    axes[1, col].set_title(
        f"{mode_name}\nB={params[0]:.2f} C={params[1]:.2f} G={params[2]:.2f}",
        fontsize=9)
    axes[1, col].axis("off")

plt.suptitle("Synthetic Pairs — Reference (top) vs Degraded (bottom)", fontsize=12)
plt.tight_layout()
plt.savefig("dataset_demo.png", dpi=120)
plt.show()
print("Saved dataset_demo.png")
````
